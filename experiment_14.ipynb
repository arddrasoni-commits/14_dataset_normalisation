```python
#Dataset
import pandas as pd
import numpy as np

data = {
    'product':['amplifier','jack','rosin','bridge'],
    'price':[10000,7000,9000,6000],
    'Units_sold':[20,30,70,50]

}

df = pd.DataFrame(data)
print(df)

```

```python
# 1. Min-Max Normalisation
# Min max normalisation for price

df['price_MinMax'] = (df['price'] - df['price'].min()) / (df['price'].max() - df['price'].min())

print('\nMin-Max Normalised price:\n')
print(df[['product','price','price_MinMax']])
```

```python
#Normalise Multiple Columns
cols=['Price','Units_Sold','Discount']

df_norm=(df[cols]-df[cols].min())/(df[cols].max()-df[cols].min())

print("Normalized Dataset")
print(df_norm)
```

```python
# 2. Z-Score normalisation
df['Units_sold_zscore'] = (df['Units_sold'] - df['Units_sold'].mean()) / df['Units_sold'].std()

print('\nZ-Score Normalised Units_sold:\n')
print(df[['product','Units_sold','Units_sold_zscore']])
```

```python
# 3. Decimal-Scaling
df['price_decimal'] = df['price'] / 100000

print('\nDecimal Scaled price:\n')
print(df[['product','price','price_decimal']])
```

# PART 2: Turning categorical variables into quantitative variables in python

```python
import pandas as pd
import numpy as np

data = {
    'order_id': [101,102,103,104,105,106],
    'gender':['male','female','female','male','female','male'],
    'payment_method':['upi','credit card','debit card','cod','upi', ' debit card'],
    'payment_category':['electronics','clothing','home','electronics','beauty','electronics'],
    'city':['pune','mumbai','delhi','bangalore','pune','hyderabad'],
    'order_value':[1528,2956,3652,2842,5962,4587]
}


df = pd.DataFrame(data)
print(df)
```

```python
# Method 1: Label Encoding
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['gender_encoded'] = le.fit_transform(df['gender'])
df
```

```python
# Method 2: One-Hot encoding
df_encoded = pd.get_dummies(df, columns=['payment_method'])

print(df_encoded)
```

```python
# Encoding Multiple Columns
df_multi = pd.get_dummies(df, columns=['payment_method','payment_category','city'])

print(df_multi)
```

```python
# Method 3: Dummy Enconding (Drop First)
df_dummy = pd.get_dummies(df, columns=['payment_method'], drop_first=True)

print(df_dummy)
```
